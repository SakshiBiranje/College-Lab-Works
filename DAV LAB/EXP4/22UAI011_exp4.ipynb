```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler  # ✅ includes StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df = pd.read_csv("input_file.csv")
```

```python
print("First 5 rows of the dataset:")
print(df.head())
```

```python
df.ffill(inplace=True)
```

```python
label_encoder = LabelEncoder()
df['Gender'] = label_encoder.fit_transform(df['Gender'])  # Male=1, Female=0
```

```python
X = df[['Age', 'Salary', 'Gender']]
y = df['Purchased']
```

```python
scaler = StandardScaler()  
X_scaled = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
acc = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)
```

```python
print("\n✅ Model Evaluation Results:")
print("Accuracy:", round(acc, 3))
print("Confusion Matrix:\n", cm)
```

```python
sample_data = pd.DataFrame({
    'Age': [30],
    'Salary': [50000],
    'Gender': [1]  # 1 = Male
})
sample_scaled = scaler.transform(sample_data)
sample_pred = model.predict(sample_scaled)
print("\nPrediction for sample input [Age=30, Salary=50000, Gender=Male]:", sample_pred[0])
```

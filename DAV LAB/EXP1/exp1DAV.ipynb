```python
import pandas as pd
```

```python
df = pd.read_csv("Employee 1000x.csv")
df.head()
```

```python
print("First 5 rows of the dataset:")
print(df.head())

print("\nData Info:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python
# Step 3: Clean the data
# Remove duplicate rows
df.drop_duplicates(inplace=True)
```

```python
# Normalize 'Salary' column if present
if 'Salary' in df.columns:
    df['NormalizedSalary'] = (df['Salary'] - df['Salary'].min()) / (df['Salary'].max() - df['Salary'].min())
```

```python
df.to_csv("Cleaned_Employee_1000x.csv", index=False)
print("\nCleaned data exported to 'Cleaned_Employee_1000x.csv'")
```

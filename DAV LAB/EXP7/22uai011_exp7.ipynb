```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score, calinski_harabasz_score
```

```python
df = pd.read_csv("documents.csv")
```

```python
print("First few documents:")
print(df.head(), "\n")
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(df['Text'])
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
agglo = AgglomerativeClustering(n_clusters=3)
```

```python
print("Clustering documents using K-Means and Agglomerative Clustering...\n")
kmeans_labels = kmeans.fit_predict(X)
agglo_labels = agglo.fit_predict(X.toarray())
```

```python
kmeans_silhouette = silhouette_score(X, kmeans_labels)
kmeans_ch = calinski_harabasz_score(X.toarray(), kmeans_labels)
agglo_silhouette = silhouette_score(X, agglo_labels)
agglo_ch = calinski_harabasz_score(X.toarray(), agglo_labels)
```

```python
print("=== K-Means Clustering ===")
print(f"Silhouette Score: {kmeans_silhouette:.4f}")
print(f"Calinski-Harabasz Index: {kmeans_ch:.4f}\n")

print("=== Agglomerative Clustering ===")
print(f"Silhouette Score: {agglo_silhouette:.4f}")
print(f"Calinski-Harabasz Index: {agglo_ch:.4f}\n")
```

```python
if kmeans_silhouette > agglo_silhouette:
    print("✅ K-Means performed better based on Silhouette Score.")
else:
    print("✅ Agglomerative Clustering performed better based on Silhouette Score.")
```
